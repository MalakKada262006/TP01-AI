```python
import nltk
import sklearn
import spacy

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")

nlp = spacy.load("fr_core_news_sm")
print("Tout est bien installé et chargé !")
```

```python
import string

text = "Bonjour Malak ! Le travail de TP (Travail Pratique) sera déposé lundi 05/10/2026, d'accord ?"

# 1. Transformation en minuscules
text_lowercase = text.lower()

# 2. Suppression de la ponctuation (avec la bibliothèque string)
text_clean = text_lowercase.translate(
    str.maketrans("", "", string.punctuation + "’«»")
)

print("Texte original :", text)
print("Texte nettoyé  :", text_clean)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# 1. Tokenisation (découpage en mots)
tokens = word_tokenize(text_clean, language="french")
print("Tokens originaux :", tokens)

# 2. Récupération des stopwords en français
french_stopwords = set(stopwords.words("french"))

# 3. Choix d'un mot supplémentaire à ajouter aux stopwords (ex: 'vraiment')
custom_stopword = "vraiment"
french_stopwords.add(custom_stopword)

# 4. Filtrage des stopwords
filtered_tokens = [
    token for token in tokens if token.lower() not in french_stopwords
]

print("\nListe des stopwords (extrait) :", list(french_stopwords)[:10])
print("\nTokens restants après filtrage :", filtered_tokens)
```

```python
proportion_supprimee = (len(tokens) - len(filtered_tokens)) / len(tokens) * 100
print(f"Proportion supprimée : {proportion_supprimee:.2f}%")
```

```python
import spacy

# Reconstitution d'une chaîne avec les tokens filtrés pour spaCy
text_filtered_str = " ".join(filtered_tokens)
doc = nlp(text_filtered_str)

# Application de la lemmatisation
lemmas = [token.lemma_ for token in doc]

# Comparaison des mots avant et après lemmatisation
print(f"{'Mot original':<20} -> {'Lemme':<20}")
print("-" * 43)
for original, lemma in zip(filtered_tokens, lemmas):
    print(f"{original:<20} -> {lemma:<20}")
```
